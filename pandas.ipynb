describe method is used to find the summary of descriptive analysis of numerical cols in our dataframe

```python
data = {
    "name" : ['ram' , 'shyam' , 'ghanshyam' , 'dhanshyam' , 'aditi' , 'jagdish' , 'raj' , 'simran'] ,
    "age" : [28 ,34 ,22,30,29,40 ,25 ,32],
    "salary" : [50 ,60, 45 , 52 , 49 , 70 , 48 , 58],
    "performance" : [85 ,90 ,78 , 92 , 88 , 95 , 80 , 89]
}
df = pd.DataFrame(data)
print("sample data")
print(df)
print("descriptive stats")
print(df.describe())
#count -- the no of non missing value in each col
#mean -- avg of the data/values in each cols
#std - means standard deviation -- tells us how muxch values in a col or are spread out -- small std means the values are close to mean.. large std means the values are very far from mean
#min --find the minimum value of the column
# 25%-- it means the first quarter of the data..it means if the data is sorted then the smallest 25% data will be below it..same goes for 50%(called median) and 75%
#max -- find the max value of the col
```

filtering rows based on conditon

```python
high_salary = df[df["salary"] > 45]
print("employees with salary > 45")
print(high_salary)

#filtering rows salary > 45 and age > 30
print("employees with age > 30 and salary > 45")
filtered = df[(df['age'] > 30) & (df['salary'] > 45)]
print(filtered)

#performace > 85 or age >30 -- either one is true (using or)
filtered_or = df[(df['age'] > 30) | (df['performance'] > 80)]
print("employees with age > 30 or performance > 85")
print(filtered_or)
```
