handling missing data

```python
#jabh kisi dataset me koi value missing hoti hnn usse missing data kehte hnn
#2types of missing data -- 1.NAN 2.None(for object data types)
# isnull() -- creates bool value -- true means nan and false means value is present
data = {
    "name": ['ram', "shyam", 'ghanshyam', 'dhanshyam', 'aditi', 'jagdish', 'raj', 'simran'],
    "age": [28, None, 22, 32, 29, 40, 25, 32],
    "salary": [50, None, 45, 52, 49, 70, 48, 58],
    "performance": [85, None, 78, 92, 88, 95, 80, 89]
}
df = pd.DataFrame(data)
print(df.isnull())
#to find which col has how many missing values
print(df.isnull().sum())
#true means data is null
```

interploation -- it is a technique in which missing values me estimated value insert kar skte hnn

eg data is 10 20 nan 40 50..

so pandas will estimate nan as 30..

interpolation is used for numeric values

uses--
1.preserve data integrity

2.smooth trends

3.avoid data loss

interpolate() method can we used to find missing values


```python
data = {
    "name": ['ram', "shyam", 'ghanshyam', 'dhanshyam', 'aditi', 'jagdish', 'raj', 'simran'],
    "age": [28, None, 22, 32, 29, 40, 25, 32],
    "salary": [50, None, 45, 52, 49, 70, 48, 58],
    "performance": [85, None, 78, 92, 88, 95, 80, 89]
}
df = pd.DataFrame(data)
#there are many interpolation methods..some important are linear, polynomial and time

df[['age', 'salary', 'performance']] = df[['age', 'salary', 'performance']].interpolate(method='linear')
print(df)
#axis =0 means rows me axis =1 means columns me
```

```python
data2 = {
    "time" : [1,2,3,4,5],
    "value" : [10 ,None , 30 , None ,50]
}
df2 = pd.DataFrame(data2)
print("before interpolation")
print(df2)

df2['value'] = df2['value'].interpolate(method = "linear")
print("after interpolation")
print(df2)

#when we are using time series data we can use interpolation..similarly with numeric data with trends we can use interpolation
```

sorting and aggregation

```python

#sorting means arranging rows on the basis of one or more values

#sorting data in one column
#sort_values()
#synatx -- df.sort_values(by = "columnname",true/false . inplace = True)
#true means ascending order and false means descending order
data3 = {
    "name" : ['arun' , 'varun' , 'karun'],
    "age" : [28 , 34 , 22],
    "salary" :[100 , 200 , 300]
}
df3 = pd.DataFrame(data3)
df3.sort_values(by = "age" , ascending=False, inplace=True)
print("sorting age by descending")
print(df3)

```

sorting by multiple columns

```python
df3.sort_values(by = ["age" , "salary"] , ascending = [True , False] , inplace = True)
print(df3)
#sorting is done first on the basis of age and if two rows have same age then on the basis of salary
```

summary statistics

```python
#syntax -- df["column name"].mean()
#.sum()
#.max()
#.min()

avg_salary = df3["salary"].mean()
print(avg_salary)
```

grouping in pandas mtlb grp me divide krna apne data ko


```python
#we can divide data in column wise
data3 = {
    "name" : ['arun' , 'varun' , 'karun' , 'narun' , 'marun'],
    "age" : [28 , 34 , 22 , 34 , 28],
    "salary" :[500 , 600 , 450 , 520 , 480]
}
df3 = pd.DataFrame(data3)
grouped = df3.groupby("age")["salary"].sum()
print(grouped)

"""
first the data is grouped by age then then summed up on the basis of salary
"""
```
