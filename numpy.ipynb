```python
py_list = [1,2,3]
print("python list multipiclation" , py_list*2)#doubles the element
np_array = np.array([1,2,3]);
print("python array multipiclation" , np_array*2) #;scalar multipiclation
# #element wise multipiclation
#see the difference between list multi and array multi

import time
start = time.time()
py_list = [i*2 for i in range(1000000)]
print("\n list operation time ", time.time() - start)

start = time.time()
#arange is a method to creat an array in an order
np_array = np.arange(1000000) *2
print("\n array operation time ", time.time() - start)
#numpy array is faster than lists
```

array reshaping

```python
arr = np.arange(12)
print("original array \n" , arr)
reshaped = arr.reshape((3,4))
print("original array \n" , reshaped)

flattened = reshaped.flatten()
print("flattened array \n" , flattened)
#ravel returns view instead of copy
#Changing the raveled array's elements can change the original array.Creating the view doesn't change the original. Modifying the view can change the original.
raveled = reshaped.ravel()
print("raveled array \n" , raveled)

#transpose
transpose = reshaped.T
print("transpose array \n" , transpose)
```

sorting array

```python
unsorted = np.array([3,1,4,1,5,9,2,6])
print("sorted array \n" , np.sort(unsorted))
arr_2d_sorted = np.array([[3,1] , [1,2] ,[2,3]])
print("sorted 2d array by col \n" , np.sort(arr_2d_sorted , axis =0))
print("sorted 2d array by row \n" , np.sort(arr_2d_sorted , axis =1))
#axis = 0 sort each column of the 2d matrix independently
#axis = 1 sort each row of the 2d matrix independently
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])
print("=== zomato sales analysis ===")
print("\n sales data shape" , sales_data.shape)
#q1 -- print sample data for first three restaurant
print(sales_data[:3])
#predict output
print(sales_data[: , 1:]) #mtlb sare rows le lo bcoz first parameter is : and column ka pehla element nii print krna


#ques3 : print the total sale
print(np.sum(sales_data , axis =0))
#np.sum is the method to find sum of all elements of a array
yearly_total = np.sum(sales_data[: , 1:] , axis =0)
print(yearly_total)
total_sum = np.sum(yearly_total)
print(total_sum)


#ques4 --minimum sales per restaurant
min_sales = np.min(sales_data[: , 1:] , axis =1)
print(min_sales)

#ques5 -- maximum sales per year

max_sales = np.max(sales_data[: , 1:] , axis =0)
print(max_sales)


#average sales per rest
avg_sales = np.mean(sales_data[: , 1:] , axis =1)

#to find the cumulative sale
cumsum = np.cumsum(sales_data[: , 1:] , axis =1)
print(cumsum)

#plotting on matplotlib
#step1 : figure ka size declare karo
plt.figure(figsize = (10 ,6))
#step2 : plot karo-- pass the array
plt.plot(np.mean(cumsum , axis =0))
plt.title("average cumulative sales per year across all restaurant")
plt.xlabel("years")
plt.ylabel("sales")
plt.grid(True)
plt.show()
```

np.sum to find the sum and np.min to find the minimum and np.max to find the maximum..and mean for avg..also we can use np.cumsum to find the cumulative value.. also in 2d array we use axis -- 1 for col and 0 for row


some vector operations

```python
vector1 = np.array([1,2,3,4,5])
vector2 = np.array([6,7,8,9,10])
print("vector addition" , vector1 + vector2)
print("\n multipiclation " , vector1 * vector2)
print("\n Dot product" , np.dot(vector1, vector2))
angle = np.arccos(np.dot(vector1 , vector2) /( np.linalg.norm(vector1) * np.linalg.norm(vector2)))
print(angle)

#to find the magnitude of a vector we use np.linalg.norm(vector1)
```
