```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import classification_report
```

```python
url = "https://raw.githubusercontent.com/611noorsaeed/100-days-Scikit-Learn-Tutorials-/refs/heads/main/6%20advertising.csv"
df = pd.read_csv(url)
df.head()
```

```python
#supervised means jisme labeled data ho...
# supervised is of two types -- regression and classification
sns.pairplot(df) #it is a function used to visualize the relationship between multiple numerical columns in a dataframe..
```

machine learning is mainly of two types --
1.supervised learning -- in it the model is trained using data where the correct answer is already given
supervised learning is of two types -- regression(in which target is a continous numerical value) and classification(in which target is a category or class)
2.unsupervised learning -- the data has no target/label.the model tries to discover patterns or structures in the data by itself
different types of unsupervised include -- clustering and dimensionality reduction(pca- principal component analysis)

making training and testing data--
first we will make training and testing data before applying linear regression

```python
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(df.iloc[: ,:-1] , df.iloc[: ,-1] , test_size = 0.2 , random_state = 42)
x_train.shape , x_test.shape , y_train.shape , y_test.shape
```

applying linear regression

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train, y_train)
y_pred = lr.predict(x_test)
print(y_pred)
```

metrices R2_score

```python
from sklearn.metrics import r2_score , mean_absolute_error
k = mean_absolute_error(y_test , y_pred)
print(k)
```

```python
r2_score(y_test, y_pred)
#r2 is coefficient of determination
```

linear regression--
it is a machine learning algo explains the relationship between the dependent(output) variable and the independent variable using a straight line..
     y = b0 + b1x
we will find the best fit line
residual is the difference between the actual value and the predicted value..

predictive system

```python
def predict_sales(tv_budget , radio_budget , newspaper_budget ):
    #the function takes a input and convert it into numpy array
    features = np.array([[tv_budget , radio_budget , newspaper_budget]])
    results = lr.predict(features).reshape(1,-1) #make 1 row and numpy automatically calculates how many columns are needed
    return results[0]
```

```python
df.iloc[0]
tv_budget = 230.1
radio_budget = 37.8
newspaper_budget = 69.2
print("actual value is" , df.iloc[0]['Sales'])
sales = predict_sales(tv_budget , radio_budget , newspaper_budget)
print("expected value is " , sales)
```
