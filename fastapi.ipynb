fastapi is a modern , high performance web framework for building apis with python..fastapi is made of 2 python libraries -- starlette and pydantic..when we build a api,api receives http request from client and then give http response thorugh http..this is managed by starlette..starlette manages how your api receives requests and sends back responses..pydantic is used to check if the data coming into your api is correct and in the right format...

philosophy of fastapi

the old frameworks in which we can create api are slow(performance issues) and it is difficult to write code in older frameworks..so thats why fastapi is created..fastapi is fast to run and fast to code the api..thats why its name is fast...every api has two components--"api code" and second is "web server"..api code helps to implement logic..web server helps to process incoming request from client..
